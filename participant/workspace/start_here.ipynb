## Workspace readiness

Run the next cell to confirm your CTFd identity, workspace, and exercise evaluators are ready. It also gives you a direct link to open, stop, or reset your workspace. It does not submit an image or change your score.


```python
from range_device import select_device, describe_device
device = select_device()
print("Compute device:", describe_device(device))

import html
import os
from urllib.request import urlopen
from IPython.display import HTML, display

ctfd_username = os.environ.get("CTFD_USERNAME", "").strip()
ctfd_user_id = os.environ.get("CTFD_USER_ID", "").strip()
ctfd_public_url = os.environ.get("CTFD_PUBLIC_URL", "http://localhost:8001").rstrip("/")
workspace_id = os.environ.get("WORKSPACE_ID", "").strip()
identity_ok = bool(ctfd_username and ctfd_user_id.isdecimal() and int(ctfd_user_id) > 0)

rows = []
if identity_ok:
    identity_status = "✅ Connected"
    identity_detail = f"{ctfd_username} (ID {ctfd_user_id})"
else:
    identity_status = "⚠️ Not attributed"
    identity_detail = "Open this workspace from CTFd → Launch Workspace."
rows.append(("CTFd account", identity_status, identity_detail))

workspace_ok = bool(workspace_id)
rows.append((
    "Workspace",
    "✅ Identified" if workspace_ok else "⚠️ Unknown",
    workspace_id or "Workspace ID was not provided to this notebook.",
))

def check_health(label, base_url, optional=False):
    health_url = f"{base_url.rstrip('/')}/health"
    try:
        with urlopen(health_url, timeout=5) as response:
            ok = response.status == 200
            status = "✅ Available" if ok else f"⚠️ HTTP {response.status}"
            detail = health_url
    except Exception as exc:
        ok = False
        status = "ℹ️ Not enabled (optional)" if optional else "❌ Unavailable"
        detail = f"{health_url} ({type(exc).__name__})"
    rows.append((label, status, detail))
    return ok

patchguard_target_ok = check_health(
    "PatchGuard evaluator",
    os.environ.get("PATCHGUARD_TARGET_URL", "http://patchguard-target:8000"),
)
arena_ok = check_health(
    "X-Ray Arena evaluator",
    os.environ.get("ARENA_URL", "http://arena-service:5000"),
)
xray_redblue_ok = check_health(
    "X-Ray Red/Blue evaluator (optional)",
    os.environ.get("XRAY_REDBLUE_URL", "http://xray-redblue:5000"),
    optional=True,
)

llm_safety_ok = check_health(
    "Breaking LLMs evaluator (optional, single-student)",
    os.environ.get("LLM_SAFETY_URL", "http://llm-safety-target:8000"),
    optional=True,
)

html_rows = "".join(
    "<tr><th>{}</th><td>{}</td><td>{}</td></tr>".format(
        html.escape(label), html.escape(status), html.escape(detail)
    )
    for label, status, detail in rows
)
manage_url = html.escape(f"{ctfd_public_url}/my-workspace", quote=True)
summary = (
    f'<p><a href="{manage_url}" target="_blank" rel="noopener" '
    'style="display:inline-block;padding:8px 14px;border-radius:6px;background:#1769aa;color:white;text-decoration:none;font-weight:600">'
    'Manage my workspace</a></p>'
    f"<h3>Workspace readiness</h3><table><tbody>{html_rows}</tbody></table>"
)
if not (identity_ok and workspace_ok and patchguard_target_ok and arena_ok):
    summary += "<p><strong>Some checks need attention.</strong> Ask your instructor if the workspace or an evaluator is unavailable.</p>"
display(HTML(summary))
```
